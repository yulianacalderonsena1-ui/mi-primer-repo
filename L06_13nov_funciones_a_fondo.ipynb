# 📚 Parte 1 · `return` a fondo: mostrar no es devolver

Usted ya escribe funciones. Hoy se trata de entender **bien** qué pasa cuando una función termina, porque ahí está el error más común de todos.

🏧 Piense en un **cajero automático**. Usted le pide plata y el cajero puede hacer dos cosas muy distintas:

- **Mostrarle en la pantalla** «Su saldo es $50.000». Usted lo ve, pero no se lleva nada.
- **Entregarle los billetes**. Ahora la plata es suya: la guarda, la gasta, la suma con otra.

En una función, `print` es la **pantalla** y `return` son los **billetes**.

| | `print(dato)` | `return dato` |
|---|---|---|
| ¿Qué hace? | Lo muestra en pantalla | Se lo **entrega** a quien llamó la función |
| ¿Se puede guardar en una variable? | No | Sí: `x = funcion()` |
| ¿Se puede seguir usando (sumar, comparar)? | No | Sí |
| ¿La función sigue después? | Sí | **No: `return` termina la función ahí mismo** |

Una función que **no tiene `return`** también entrega algo: entrega `None` (el «nada» que conoció ayer).

````python
def area_mostrar(base, altura):
    print(base * altura)            # solo lo muestra

def area(base, altura):
    return base * altura            # lo entrega

x = area_mostrar(4, 5)              # aquí se imprime 20
print(x)                            # pero x quedó con None

y = area(4, 5)                      # aquí no se imprime nada
print(y)                            # y sí tiene el 20
print(y * 2)                        # y se puede seguir usando
````

**Qué pasó, línea por línea**
- `x = area_mostrar(4, 5)`: la función imprimió `20`, terminó sin `return` y por eso a `x` le llegó `None`.
- `y = area(4, 5)`: la función no imprimió nada; **entregó** el 20 y quedó guardado en `y`. Por eso después se puede multiplicar.

> **Regla de oro:** una función que **calcula** algo lo devuelve con `return`. El `print` lo decide quien la llama.

### `return` termina la función
Apenas Python llega a un `return`, **sale de la función**, aunque queden líneas debajo o vueltas del ciclo sin dar. Eso sirve para responder apenas se sabe la respuesta (así funciona el `buscar` de ayer):

```python
def clasificar(nota):
    if nota >= 4.5:
        return "excelente"      # si entra aquí, sale: lo de abajo no se mira
    if nota >= 3.0:
        return "aprobado"
    return "reprobado"          # solo llega aquí si no salió antes
```

⚠️ Y es también la trampa: un `return` **dentro de un `for`**, sin un `if` que lo proteja, hace que la función salga **en la primera vuelta**.

### Funciones que responden sí o no
Una comparación ya vale `True` o `False`, así que se puede devolver directamente:

```python
def es_mayor_de_edad(edad):
    return edad >= 18           # devuelve True o False
```

Y se usa donde iría una condición: `if es_mayor_de_edad(edad):`. Por costumbre, estas funciones se llaman `es_…`, `tiene_…` o `puede_…`.

````python
def clasificar(nota):
    if nota >= 4.5:
        return "excelente"
    if nota >= 3.0:
        return "aprobado"
    return "reprobado"

def es_mayor_de_edad(edad):
    return edad >= 18

print(clasificar(4.8))
print(clasificar(3.2))
print(clasificar(1.0))
print(es_mayor_de_edad(17))
if es_mayor_de_edad(20):
    print("Puede entrar")
````

# 📚 Parte 2 · Parámetros con valor por defecto

Usted lleva semanas **usando** esto sin saberlo:

```python
print("Hola", end="")                 # end tiene un valor por defecto: el salto de línea
sorted(lista, reverse=True)           # reverse por defecto es False
nombre.split()                        # si no le dice con qué partir, parte por espacios
```

Esas funciones tienen parámetros que **ya traen un valor** por si usted no dice nada. Hoy aprende a escribir las suyas así.

🍔 Es como pedir una hamburguesa: si usted no dice nada, llega **con todo**. Solo habla si quiere cambiar algo: «sin cebolla».

```python
def saludar(nombre, saludo="Hola"):
    return f"{saludo}, {nombre}"
```

- `nombre` es **obligatorio**: no tiene valor por defecto.
- `saludo` es **opcional**: si no lo mandan, vale `"Hola"`.

### Tres maneras de llamarla
```python
saludar("Camila")                          # usa el valor por defecto
saludar("Mateo", "Buenas tardes")          # por posición: el segundo dato es el saludo
saludar("Sofía", saludo="Qué más")         # por nombre: digo cuál parámetro estoy mandando
```

Llamar **por nombre** sirve sobre todo cuando hay varios opcionales y solo quiere cambiar uno: se salta los demás.

> **Regla:** en el `def`, los parámetros con valor por defecto van **al final**, después de los obligatorios. `def f(a=1, b):` da error.

````python
def saludar(nombre, saludo="Hola"):
    return f"{saludo}, {nombre}"

print(saludar("Camila"))
print(saludar("Mateo", "Buenas tardes"))
print(saludar("Sofía", saludo="Qué más"))

def precio_final(precio, descuento=0, iva=19):
    con_descuento = precio - precio * descuento // 100
    return con_descuento + con_descuento * iva // 100

print(precio_final(100000))                        # sin descuento, IVA del 19
print(precio_final(100000, 10))                    # 10 de descuento, IVA del 19
print(precio_final(100000, iva=5))                 # sin descuento, IVA del 5
print(precio_final(100000, descuento=20, iva=0))   # 20 de descuento, sin IVA
````

**Qué pasó, línea por línea**
- `precio_final(100000)`: solo llegó el precio. `descuento` quedó en 0 e `iva` en 19 → 119000.
- `precio_final(100000, 10)`: el segundo dato cae, **por posición**, en `descuento` → 90000 más su IVA = 107100.
- `precio_final(100000, iva=5)`: se **saltó** el descuento nombrando el IVA → 105000.

# 📚 Parte 3 · Lo que pasa en la función, se queda en la función

🍳 Una función es como la **cocina de un restaurante**. Usted, desde la mesa, entrega el pedido (los **parámetros**) y recibe el plato (el **`return`**). Lo que pasa adentro —las ollas, los ingredientes, las cuentas— **no sale de la cocina**, y cuando el plato sale, la cocina se limpia.

Las variables que se crean dentro de una función se llaman **locales**: nacen cuando la función empieza y **desaparecen** cuando termina.

````python
def calcular_iva(precio):
    iva = precio * 19 // 100        # iva nace aquí adentro
    return iva

resultado = calcular_iva(50000)
print(resultado)                    # 9500: el dato salió por el return
print(iva)                          # ❌ afuera, la variable iva no existe
````

**Qué pasó:** se imprimió `9500` y después salió `NameError: name 'iva' is not defined`. La variable `iva` vivía **dentro** de `calcular_iva` y desapareció al terminar. El dato sí salió, pero por la única puerta que hay: el `return`, que lo dejó en `resultado`.

### Un parámetro es una variable local más
Si adentro le cambia el valor a un parámetro, cambia **la variable de adentro**; la de afuera sigue igual, aunque se llamen lo mismo:

````python
def subir_precio(precio):
    precio = precio + 1000          # cambia el precio DE ADENTRO
    return precio

precio = 5000
nuevo = subir_precio(precio)
print(precio)                       # el de afuera sigue en 5000
print(nuevo)                        # el resultado llegó por el return

def agregar_pan(lista):
    lista.append("pan")             # no crea nada: modifica LA MISMA lista que le pasaron

compras = ["leche"]
agregar_pan(compras)
print(compras)
````

**Qué pasó, línea por línea**
- `precio = precio + 1000` dentro de la función **arma un dato nuevo** y lo guarda en el `precio` de adentro. El de afuera no se entera: sigue en 5000.
- Con la lista es distinto: `.append()` no arma una lista nueva, **modifica la que le pasaron**, que es la misma de afuera. Es lo que vio ayer: `buscar` entrega la misma ficha, y al cambiarla cambia la lista.

> **Las dos reglas del buen funcionario**
> 1. Lo que la función necesita, le entra por los **parámetros**. No use adentro variables de afuera (la única excepción son las tablas fijas en MAYÚSCULAS, como `TARIFAS`).
> 2. Lo que la función produce, sale por el **`return`**.
>
> Una función así se puede probar sola, se puede copiar a otro programa y nunca daña nada por accidente.

# 📚 Parte 4 · Devolver dos datos y funciones que usan funciones

### Devolver más de un dato
A veces una función averigua **dos cosas** en el mismo recorrido (el menor y el mayor, el total y el conteo). Se devuelven separadas por coma, y quien llama las recibe en **dos variables**, en el mismo orden:

```python
return menor, mayor              # dentro de la función

bajo, alto = extremos(ventas)    # quien llama: dos variables, mismo orden
```

````python
def extremos(numeros):
    menor = numeros[0]
    mayor = numeros[0]
    for n in numeros:
        if n < menor:
            menor = n
        if n > mayor:
            mayor = n
    return menor, mayor

ventas = [120, 80, 200, 150]
bajo, alto = extremos(ventas)
print("La más baja:", bajo)
print("La más alta:", alto)
````

**Qué pasó:** la función recorrió la lista **una sola vez** y entregó dos datos. `bajo` recibió el primero (80) y `alto` el segundo (200). Si pone una sola variable a la izquierda, recibe los dos datos juntos entre paréntesis (eso se llama *tupla* y es el tema de la próxima guía).

### Funciones que usan funciones
Un problema grande se parte en **problemas pequeños**, cada uno con su función, y después se arman como piezas. Cada función hace **una sola cosa** y la hace bien:

````python
def promedio(notas):
    return sum(notas) / len(notas)

def estado(nota):
    if nota >= 3.0:
        return "aprobó"
    return "no aprobó"

def reporte(nombre, notas):
    prom = promedio(notas)                           # usa la primera
    return f"{nombre}: {prom:.1f} ({estado(prom)})"  # y la segunda

print(reporte("Camila", [4.0, 3.5, 4.5]))
print(reporte("Mateo", [2.0, 3.0, 2.5]))
````

**Qué pasó:** `reporte` no sabe sacar promedios ni decidir quién aprueba: **les pregunta** a las otras dos. Si mañana cambia la nota para aprobar, se toca **una sola línea**, dentro de `estado`.

> **Señal de que hace falta una función:** si usted copió y pegó un bloque de código y solo le cambió un nombre, ese bloque quiere ser una función y lo que cambió quiere ser un parámetro.

### Ejercicio 6 · El programa repetido  
**Nivel:** ⭐⭐ Medio · **Tipo:** modificar un programa que funciona  
**👥 En pareja:** B pilotea (escribe), A copilotea (dicta y revisa)

**📖 La situación:** este programa **funciona**: saca el total y la mejor venta del lunes y del martes. Pero quien lo escribió **copió y pegó** el mismo bloque dos veces. Ahora llegaron las ventas del miércoles y no queremos pegarlo una tercera vez.

**📝 Lo que debe hacer, paso a paso:**
1. Ejecute el programa tal como está y anote en papel lo que imprime.
2. **Papel primero:** compare los dos bloques. ¿Qué es **igual** en los dos? ¿Qué es lo **único** que cambia? Lo que cambia será el parámetro.
3. Escriba una función `resumen(ventas)` que haga ese trabajo y **devuelva dos datos**: el total y la mejor venta. La función no imprime.
4. Reemplace cada bloque repetido por **una llamada** a la función (recibiendo los dos datos) y su `print`.
5. Agregue el miércoles: `ventas_miercoles = [50000, 300000]`, con su llamada y su línea.
6. Las líneas del lunes y el martes deben salir **idénticas** a las del paso 1.

**✅ Compruebe antes de seguir:**
- ☐ El ciclo `for` quedó escrito **una sola vez** en todo el programa.
- ☐ `resumen` no usa `ventas_lunes` ni ninguna otra variable de afuera: solo su parámetro.
- ☐ Agregar el jueves serían solo tres líneas más.

**👀 Así debe verse al ejecutarlo**:
```
Lunes: total $400000, mejor venta $200000
Martes: total $370000, mejor venta $150000
Miércoles: total $350000, mejor venta $300000
```

<details><summary>💡 Ver pista (último recurso: primero intente y relea la explicación)</summary>

La función es el bloque repetido, cambiando `ventas_lunes` por el parámetro `ventas` y terminando en `return total, mayor`. Cada día queda en dos líneas: `total, mayor = resumen(ventas_lunes)` y su `print`.

</details>

````python
ventas_lunes = [120000, 80000, 200000]
ventas_martes = [90000, 150000, 60000, 70000]

total = 0
mayor = ventas_lunes[0]
for venta in ventas_lunes:
    total = total + venta
    if venta > mayor:
        mayor = venta
print(f"Lunes: total ${total}, mejor venta ${mayor}")

total = 0
mayor = ventas_martes[0]
for venta in ventas_martes:
    total = total + venta
    if venta > mayor:
        mayor = venta
print(f"Martes: total ${total}, mejor venta ${mayor}")
````
